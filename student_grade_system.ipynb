```python
# Student Grade Management System
# This program collects student names and marks, calculates averages, assigns grades, and displays a report

print("Welcome to the Student Grade Management System!\n")
```

```python
# Ask how many students to enter
num_students = int(input("Enter number of students: "))
```

```python
# Initialize list to store student data
students = []  # Each student will be stored as a dictionary

# Loop to collect each student's data
for i in range(1, num_students + 1):
    print(f"\nEnter details for student {i}:")
    
    # Get student's name
    name = input(f"Enter name of student {i}: ").strip()
    
    # Collect marks for three subjects
    while True:
        try:
            math = float(input("Enter marks for Math: "))
            science = float(input("Enter marks for Science: "))
            english = float(input("Enter marks for English: "))


             # Validate marks are between 0 and 100
            if all(0 <= mark <= 100 for mark in [math, science, english]):
                break
            else:
                print("Marks must be between 0 and 100. Try again.")
        except ValueError:
            print("Please enter valid numbers for marks.")
    
    # Calculate average
    average = (math + science + english) / 3
    
    # Determine grade based on average
    if 80 <= average <= 100:
        grade = "A"
    elif 70 <= average < 80:
        grade = "B"
    elif 60 <= average < 70:
        grade = "C"
    elif 50 <= average < 60:
         grade = "D"
    else:
        grade = "F"
    
    # Store student data in a dictionary and add to the list
    student = {
        "name": name,
        "average": average,
        "grade": grade
    }
    students.append(student)
            
```

```python
# Display the student report
print("\n--- Student Report ---")
print(f"{'Name':<10} {'Average':<8} {'Grade':<5}")
print("-" * 26)

# Track total for class average
total_average = 0

for student in students:
    print(f"{student['name']:<10} {student['average']:<8.1f} {student['grade']:<5}")
    total_average += student['average']

print("-" * 26)
```

```python
# Calculate and display class average
class_average = total_average / num_students
print(f"Class Average: {class_average:.1f}%")
```
